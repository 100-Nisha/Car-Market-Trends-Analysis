```python
# 17. Can we finfd out data of only two wheelers from this data?

bike_keywords = ['Royal Enfield', 'KTM', 'Bajaj', 'Honda CB', 'Yamaha', 'TVS', 'Hero', 'Activa', 'Suzuki Access', 'Hyosung', 'UM Renegade', 'Activa']
bikes_data = data[data['Car_Name'].str.contains('|'.join(bike_keywords), case=False, na=False)]
print(f"Q17: Extracted {bikes_data.shape[0]} two-wheeler records.")
```

```python
# Q20. which is the most bike here?

most_sold_bike = bikes_data['Car_Name'].value_counts().idxmax()
print(f"Q20: Most sold bike: {most_sold_bike}")
```

```python
# Q21. Do you find any in two wheelers which exceeded the general expectation? can you find reason for it?

# 1. Isolate the Two-Wheelers section safely using .copy()
bike_keywords = ['Royal Enfield', 'KTM', 'Bajaj', 'Honda CB', 'Yamaha', 'TVS', 'Hero', 'Activa', 'Access', 'Hyosung', 'UM Renegade']
bikes_data = data[data['Car_Name'].str.contains('|'.join(bike_keywords), case=False, na=False)].copy()

# 2. Calculate the Value Retention Rate metric
bikes_data['Retention_Rate'] = (data['Selling_Price'] / bikes_data['Present_Price']) * 100

# 3. Sort to find the absolute best transaction deal
top_bike = bikes_data.sort_values(by='Retention_Rate', ascending=False).iloc[0]

print("--- Q21: EXCEPTIONAL TWO-WHEELER DEAL ---")
print(f"Vehicle Model:    {top_bike['Car_Name']}")
print(f"Year / Mileage:   {top_bike['Year']} | {top_bike['Kms_Driven']} km")
print(f"Value Retention:  {top_bike['Retention_Rate']:.2f}%")
```

```python
# Q22. Can we find out data of only cars from this data?

cars_data = data[data.index.isin(bikes_data.index)]
print(f"Q22: Extracted {cars_data.shape[0]} car records.")
```

```python
# Q23. Which is the oldest car sold here?

oldest_car = cars_data.loc[cars_data['Year'].idxmin()]
print(f"Q23: Oldest car: {oldest_car['Car_Name']} ({oldest_car['Year']})")
```

```python
# Q24. Which is the newest car sold here?

newest_car = cars_data.loc[cars_data['Year'].idxmax()]
print(f"Q24: Newest car: {newest_car['Car_Name']} ({newest_car['Year']})")
```

```python
# Q25. Do you find any deal in cars which exceeded the general excepectation? can you find reason for it?

# 1. Define two-wheeler criteria to exclude them accurately
bike_keywords = ['Royal Enfield', 'KTM', 'Bajaj', 'Honda CB', 'Yamaha', 'TVS', 'Hero', 'Activa', 'Access', 'Hyosung', 'UM Renegade']
bikes_mask = data['Car_Name'].str.contains('|'.join(bike_keywords), case=False, na=False)

# 2. Filter using the inverse mask (~) to leave only the Car records
cars_data = data[bikes_mask].copy()

print("--- Q25: CARS FILTER SUMMARY ---")
print(f"Total Rows in Original Dataset: {len(data)}")
print(f"Total Extracted Car Records:    {len(cars_data)}")

# Display the first few car records to confirm
print("\nFirst 5 Rows of the Cars Subset:")
print(cars_data[['Car_Name', 'Year', 'Selling_Price']].head())
```
